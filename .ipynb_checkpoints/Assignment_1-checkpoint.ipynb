```python
#Area Of Circle#

pi=eval(input("enter the pi:"))
radius=eval(input("enter the value of radius:"))  
area_of_circle=pi*radius*radius
print(f"the area of circle is:",area_of_circle)
```

```python
#Area of Rectangle#
length=eval(input("enter the length of rectangle:"))
breath=eval(input("enter the breath of rectangle:")  )
area=length*breath
print(f"the area of rectangle is:",area)
```

```python
##Area of squre##
length=eval(input("enter the length of squre:"))
squre=length*length
print(f"the area of squre is:",squre)
```

```python
##convert meter into cm##
meter=eval(input("enter the meter value:"))
Value=meter*100
print(f"the centimeter value is:",Value)
```

```python
#convert weight in pound to KG##
pound=eval(input("enter the value of pound:"))
Kg_value=pound*0.453592
print(f"the kg value is:",Kg_value)
```

```python
##Convert $ to rupee##
Doller=eval(input("enter the value of Doller:"))
Money=Doller*85
print(f"the indian currency amount:",Money)
```

```python
##Area of triangle##
area_base=eval(input("enter the value of base:"))
area_height=eval(input("enter the value of hight:"))
final_area=0.5*area_base*area_height
value=round(final_area,2)
print(f"the area of triangle is:",value)
```
